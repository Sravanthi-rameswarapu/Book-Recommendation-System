```python
import numpy as np

# Convert bestbooks DataFrame into a set for faster lookup
bestbooks_set = set(bestbooks['book-title'].str.strip().str.lower())  # Normalize titles

def recommend(book_name):
    # Normalize input book title to match bestbooks format
    book_name_normalized = book_name.strip().lower()

    # Convert pt.index to a list for string operations
    pt_index_list = [title.strip().lower() for title in pt.index.tolist()]

    # Check if the book is in the dataset
    if book_name_normalized not in pt_index_list:
        print("\nBook not found in dataset.")
        return

    # Fetch index of the given book
    index = np.where(np.array(pt_index_list) == book_name_normalized)[0].item()

    # Sort similarity scores in descending order
    similar_items = sorted(
        list(enumerate(similarity_scores[index])),
        key=lambda x: x[1],
        reverse=True
    )

    # Filter out books with zero similarity and ensure they exist in bestbooks
    filtered_books = [
        pt.index[i[0]] for i in similar_items 
        if i[1] > 0 and pt.index[i[0]].strip().lower() in bestbooks_set
    ][1:6]  # Skip first item (it’s the input book itself)

    if filtered_books:
        print("\nRecommended books for:", book_name)
        for book in filtered_books:
            print(book)
    else:
        print("\nNo similar books found in bestbooks for:", book_name)
```

```python

mood_to_genres = {
    'happy': ['Comedy', 'Humor', 'Adventure', 'Romance'],
    'sad': ['Tragedy', 'Drama', 'Historical Fiction'],
    'exciting': ['Thriller', 'Mystery', 'Crime', 'Action', 'Suspense', 'Fantasy', 'Science Fiction'],
    'relaxing': ['Contemporary', 'Poetry', 'Self-Help', 'Historical'],
    'thoughtful': ['Nonfiction', 'Politics', 'Reference', 'Literature', 'Philosophy'],
    'curious': ['Fantasy', 'Science Fiction', 'Alternate History', 'Mythology', 'Speculative Fiction']
}
```

```python
def map_genres_to_mood(genre_list):
    for genre in genre_list:
        for mood, genres in mood_to_genres.items():
            if genre in genres:
                return mood
    return 'other'  
```

```python
import ast
#moodbooks['genre_list'] = moodbooks['genres'].apply(lambda x: ast.literal_eval(x))
moodbooks['genre_list'] = moodbooks['genres'].apply(lambda x: x.split(", ") if isinstance(x, str) else [])
```

```python

moodbooks['mood'] = moodbooks['genre_list'].apply(map_genres_to_mood)
```
